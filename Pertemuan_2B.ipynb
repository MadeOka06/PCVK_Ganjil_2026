```python
import cv2
import matplotlib.pyplot as plt
nama_file = 'made.jpg'
img_sampel = cv2.imread('/content/drive/MyDrive/Citra/20260901_130546.jpg')
channel_b = img_sampel[:, :, 0].copy()
tinggi, lebar = channel_b.shape
x1, y1 = 10, tinggi - 70
x2, y2 = 250, tinggi - 10
cv2.rectangle(channel_b, pt1=(x1, y1), pt2=(x2, y2), color=255, thickness=3)
font = cv2.FONT_HERSHEY_SIMPLEX
ukuran_font = 0.8
warna_font = 255
ketebalan_font = 2
posisi_teks = (x1 + 10, y2 - 20)
cv2.putText(channel_b, nama_file, posisi_teks, font, ukuran_font, warna_font, ketebalan_font)
plt.imshow(channel_b, cmap='gray')
plt.title("Soal 8 & 9: Channel B dengan Teks")
plt.show()
```
